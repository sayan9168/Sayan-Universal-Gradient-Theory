## Overview

A core pillar of SUG Theory is **thermal mapping**: the thermal gradient $\tau = \Delta T / T_0$ acts in direct opposition to the energy driver:
$$\theta_S(x, \tau) = K \frac{\sqrt{e}}{\tau \ln x}$$

This notebook showcases:
1. 2D Thermal-Distance Energy Mapping (contours and colormaps).
2. Multi-source deep space simulation and spatial field superposition.
3. Interactive 3D visualization of the field surface with Plotly.


## 1. Imports and Setup

```python
import numpy as np
import matplotlib.pyplot as plt

from sayan_gradient import (
    ReferenceScales,
    Source2D,
    plot_thermal_energy_map,
    plot_deep_space_field_2d,
    plot_surface_3d,
)

sc = ReferenceScales(E0=1.0, T0=1.0, r0=1.0, K=1.0)
```

## 2. 2D Thermal-Distance Energy Map

The heatmap illustrates how cold ambient regions (small $\tau$) amplify field reach, while intense thermal gradients suppress it:

```python
fig = plot_thermal_energy_map(
    energy=1.0,
    x_range=(1.1, 10.0),
    tau_range=(0.2, 3.0),
    scales=sc,
    backend="matplotlib"
)
plt.show()
```

## 3. Multi-Source Deep Space Field Simulation

We model a cosmic sector with 3 high-energy nuclear sources positioned in deep space:

```python
sources = [
    Source2D(x=-8.0, y=5.0, energy=200.0),
    Source2D(x=6.0, y=-4.0, energy=150.0),
    Source2D(x=0.0, y=-10.0, energy=75.0),
]

fig = plot_deep_space_field_2d(
    sources=sources,
    x_bounds=(-25.0, 25.0),
    y_bounds=(-25.0, 25.0),
    grid_res=300,
    delta_T=1.0,
    scales=sc,
    backend="matplotlib",
    title="Deep Space Multi-Source Field Distribution"
)
plt.show()
```

## 4. Interactive 3D Surface Visualization with Plotly

Interactive 3D surfaces allow intuitive exploration of the geometric fall-off and thermal trade-offs:

```python
fig_plotly_3d = plot_surface_3d(
    energy=1.0,
    x_range=(1.1, 8.0),
    tau_range=(0.3, 2.5),
    grid_size=(40, 40),
    scales=sc,
    backend="plotly"
)
# In a live Jupyter notebook, call: fig_plotly_3d.show()
print("Interactive 3D Plotly Figure initialized successfully.")
print(f"Traces: {len(fig_plotly_3d.data)}, Title: {fig_plotly_3d.layout.title.text}")
```
